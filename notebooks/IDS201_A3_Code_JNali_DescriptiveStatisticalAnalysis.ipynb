## (A) Box and whisker plot for Global kilotons of CO2 Emission - Central tendency (Full view with outliers and Zoomed view showing box and whiskers)

Note: Two box and whisker plots are made for a cleaner view (Full view and zoomed view)

```python

# These are importing the tools needed
import pandas as pd  
import matplotlib.pyplot as plt
import seaborn as sns

# This loads the CVS Carbon (CO2) Emissions data into the workspace
dataset = pd.read_csv('Carbon_(CO2)_Emissions_by_Country.csv')

# This is the reference for finding the central tendency values:
# Mankar, A. (2021, June 23). Use Pandas to Calculate Statistics in Python. GeeksforGeeks. https://www.geeksforgeeks.org/python/use-pandas-to-calculate-statistics-in-python/
# This calculates mean, median and mode for CO2 emissions
mean_value = dataset['Kilotons of Co2'].mean()
median_value = dataset['Kilotons of Co2'].median()
mode_value = dataset['Kilotons of Co2'].mode()[0]
skewness = dataset['Kilotons of Co2'].skew()

# This is the reference for creating the box plot:
# Tutorialpoint. (2021). Matplotlib Box Plot. https://www.tutorialspoint.com/matplotlib/matplotlib_box_plot.htm
# This creates the boxplot from the column "Kilotons of Co2" in the dataset
fig, (ax1, ax2) = plt.subplots(2, 1, figsize=(14, 12))


# This adds the coloured lines for the central tendency values and shows their values in the legend
# These values are for the global mean, median and mode values

# This is for the full view of the box plot with the outliers
# These are the references for the full view box plots:
# Waskom, M. L. (2021). seaborn: Statistical data visualization. Journal of Open Source Software, 6(60), 3021. https://seaborn.pydata.org/generated/seaborn.boxplot.html
# kiwkandmd. (2025, July 23). Creating Multiple Boxplots on the Same Graph from a Dictionary. Geeks4geeks. https://www.geeksforgeeks.org/data-visualization/creating-multiple-boxplots-on-the-same-graph-from-a-dictionary/
sns.boxplot(data = dataset, x = 'Kilotons of Co2', y = 'Region', hue = 'Region', palette = 'Set2', ax = ax1)
ax1.axvline(mean_value, color ='red', linestyle ='--', label =f'Mean: {mean_value:,.0f}')
ax1.axvline(median_value, color ='green', linestyle ='-', label =f'Median: {median_value:,.0f}')
ax1.axvline(mode_value, color ='orange', linestyle =':', label =f'Mode: {mode_value:,.0f}')
# This is for the title of the full view box plot
ax1.set_title('Global Kilotons of CO2 Emission - Central tendency (Full View with Outliers)')
ax1.set_xlabel("Kilotons of CO2")
# This is for the legend to make sure it lies outside the box plot and does not cover any of the data points
# The is the reference for the legend:
# Bobbit, Z. (2022, April 1). How to Position a Legend in Matplotlib. Statology. https://www.statology.org/matplotlib-legend-position/
ax1.legend(title='Global Metrics', bbox_to_anchor=(1.02, 1), loc='upper left', borderaxespad=0)


# This is for the zoomed in view of the box plot showing the box and whiskers more clearly
sns.boxplot(data = dataset, x = 'Kilotons of Co2', y = 'Region', hue = 'Region', palette = 'Set2', ax = ax2)
ax2.axvline(mean_value, color ='red', linestyle ='--', label =f'Mean: {mean_value:,.0f}')
ax2.axvline(median_value, color ='green', linestyle ='-', label =f'Median: {median_value:,.0f}')
ax2.axvline(mode_value, color ='orange', linestyle =':', label =f'Mode: {mode_value:,.0f}')
# This is for the title of the zoomed in box plot
ax2.set_title('Global Kilotons of CO2 Emission - Central tendency (Zoomed in showing box and whiskers)')
ax2.set_xlabel("Kilotons of CO2")   
# This is for the legend to make sure it lies outside the box plot and does not cover any of the data points
ax2.legend(title='Global Metrics', bbox_to_anchor=(1.02, 1), loc='upper left', borderaxespad=0)

# This is for the x-axis limits to zoom in on the central tendency values and show the box and whiskers more clearly
# This is the reference for setting the x-axis limits:
# matplotlib. (2025). matplotlib.axes.Axes.set_xlim. https://matplotlib.org/stable/api/_as_gen/matplotlib.axes.Axes.set_xlim.html
ax2.set_xlim(-10000, 350000)

#This adds the key to the top right corner of the box plot

plt.tight_layout()
plt.show()

```

#### These are the central tendency values for 'Kiloton of CO2 emission' for each individual region

```python

# These are importing the tools needed
import pandas as pd

# Load your dataset
dataset = pd.read_csv('Carbon_(CO2)_Emissions_by_Country.csv')

# This is to Group by 'Region' and calculate mean, median, and mode for the kilotons of CO2 emissions
# We use a custom lambda function for Mode because a dataset can have multiple modes
# This the reference for grouping the data and calculating the mean, median, and mode:
# Pythoncentral. (2022, August 16). Pandas GroupBy Explained: A Guide on Using Pandas GroupBy. Python Central. https://www.pythoncentral.io/pandas-groupby-explained-a-guide-on-using-pandas-groupby/ 
Region_states = dataset.groupby('Region')['Kilotons of Co2'].agg([
    ('Mean', 'mean'),
    ('Median', 'median'),
    ('Skewness', 'skew'),
    ('Mode', lambda x: x.mode().iloc[0] if not x.mode().empty else None)
]).reset_index()

# This is for Formatting the output 
print("-" * 53)
print("REGIONAL CENTRAL TENDENCY ANALYSIS (Kilotons of CO2)")
print("-" * 53)


# We use an formatting string to format the numbers with commas for readability
# This is the reference for the iterrows:
# Programiz. (2025). Pandas iterrows() . https://www.programiz.com/python-programming/pandas/methods/iterrows
# This is the reference for the formatting string:
# Python Documentation. (2023). Formatted string literals. https://docs.python.org/3/tutorial/inputoutput.html
for index, row in Region_states.iterrows():
    print(f"Region: {row['Region']}")
    print(f"  - Mean:   {row['Mean']:,.2f} kt")
    print(f"  - Median: {row['Median']:,.2f} kt")
    print(f"  - Mode:   {row['Mode']:,.2f} kt")
    print(f"  - Skewness: {row['Skewness']:,.2f}")
```

##### These are the list of countries in Asia with the outliers identified (Kiloton of CO2)
Note: This code is reused for the other regions

```python

import pandas as pd

# Load your dataset
dataset = pd.read_csv('Carbon_(Co2)_Emissions_by_Country.csv')

# This isolates the region 'Asia' from the others in the dataset
asia_dataset = dataset[dataset['Region'] == 'Asia'].copy()

# This calculates the mean of the kiloton of CO2 for each country to identify the outliers then groups them
# This is the reference for summarizing, grouping the data and calculating the mean for each country in the Asia region:
# Prachi. (2025, July 23). How to reset index after Groupby pandas?. Geeksforgeeks. https://www.geeksforgeeks.org/python/how-to-reset-index-after-groupby-pandas/
country_stats = asia_dataset.groupby('Country')['Kilotons of Co2'].mean().reset_index()

# This is to calculate the IQR and its boundaries to identify the outliers
# These are the references for calculating the IQR and its boundaries:
# Programiz. (2026). Pandas quantile(). Programiz. https://www.programiz.com/python-programming/pandas/methods/quantile
# PracticalStats. (2026). How to Detect Outliers in Python: The IQR Method (Pandas Tutorial). Youtube. https://www.youtube.com/watch?v=XuoMdeTqteM
Q1 = country_stats['Kilotons of Co2'].quantile(0.25)
Q3 = country_stats['Kilotons of Co2'].quantile(0.75)
IQR = Q3 - Q1
upper_whisker = Q3 + 1.5 * IQR

# This categorizes the countries as outliers or typical base on the whiskers and IQR values
# These are the references for categorizing the countries:
# Walia, A & Kumar, P. (2022, August 4). Pandas DataFrame apply() Examples. DigitalOcean. https://www.digitalocean.com/community/tutorials/pandas-dataframe-apply-examples
country_stats['Reliability Status'] = country_stats['Kilotons of Co2'].apply(lambda x: 'OUTLIER' if x > upper_whisker else 'TYPICAL')

# This stops Pandas from hiding the countries and displays all the countries 
# This is the reference 
# Pandas. (2026). Options and settings. Pandas. https://pandas.pydata.org/docs/user_guide/options.html
pd.set_option('display.max_rows', None)  
pd.set_option('display.width', 1000)

# This is for sorting the countries by their mean kiloton of CO2 emissions and displaying their reliability status (Biggest/outliers at the top)
print(country_stats.sort_values(by='Kilotons of Co2', ascending=False)[['Country', 'Reliability Status']])
```

```python

# These are importing the tools needed
import pandas as pd

# This loads the dataset
dataset = pd.read_csv('Carbon_(CO2)_Emissions_by_Country.csv')

# This is to Group by 'Region' and calculate mean, median, and mode for metric tons per capita of CO2 emissions
# We use a custom lambda function for Mode because a dataset can have multiple modes
# This the reference for grouping the data and calculating the mean, median, and mode:
# Pythoncentral. (2022, August 16). Pandas GroupBy Explained: A Guide on Using Pandas GroupBy. Python Central. https://www.pythoncentral.io/pandas-groupby-explained-a-guide-on-using-pandas-groupby/ 
regional_stats = dataset.groupby('Region')['Metric Tons Per Capita'].agg([
    ('Mean', 'mean'),
    ('Median', 'median'),
    ('Skewness', 'skew'),
    ('Mode', lambda x: x.mode().iloc[0] if not x.mode().empty else None)
]).reset_index()

# This is for formating the output
print("-" * 60)
print("REGIONAL CENTRAL TENDENCY ANALYSIS (Metric Tons Per Capita)")
print("-" * 60)



# Formatting string are used to format the numbers with commas for readability
# This is the reference for the iterrows:
# Programiz. (2025). Pandas iterrows() . https://www.programiz.com/python-programming/pandas/methods/iterrows
# This is the reference for the formatting string:
# Python Documentation. (2023). Formatted string literals. https://docs.python.org/3/tutorial/inputoutput.html
for index, row in regional_stats.iterrows():
    print(f"Region: {row['Region']}")
    print(f"  - Mean:   {row['Mean']:,.2f} metric tons per capita")
    print(f"  - Median: {row['Median']:,.2f} metric tons per capita")
    print(f"  - Mode:   {row['Mode']:,.2f} metric tons per capita")
    print(f"  - Skewness: {row['Skewness']:,.2f}")
```

```python

# These are importing the tools needed
import pandas as pd

# This loads the dataset
dataset = pd.read_csv('Carbon_(CO2)_Emissions_by_Country.csv')

# This is to define the functions to calculate Range and IQR for the "Kilotons of Co2" column in the dataset
def calculate_range(x):
    return x.max() - x.min()

def calculate_iqr(x):
    return x.quantile(0.75) - x.quantile(0.25)

# This groups by 'Region' and calculates the range, IQR, Standard Deviation and variance for kiloton of CO2 emissions
# Pythoncentral. (2022, August 16). Pandas GroupBy Explained: A Guide on Using Pandas GroupBy. Python Central. https://www.pythoncentral.io/pandas-groupby-explained-a-guide-on-using-pandas-groupby/ 
regional_variability = dataset.groupby('Region')['Kilotons of Co2'].agg([
    ('Range', calculate_range),
    ('IQR', calculate_iqr),
    ('Std_Dev', 'std'),
    ('Variance', 'var')
]).reset_index()

# Formatting the output for professional presentation
print("-" * 53)
print("REGIONAL VARIABILITY ANALYSIS (Kilotons of CO2)")
print("-" * 53)


# Formatting string are used to format the numbers with commas for readability
# This is the reference for the iterrows:
# Programiz. (2025). Pandas iterrows() . https://www.programiz.com/python-programming/pandas/methods/iterrows
# This is the reference for the formatting string:
# Python Documentation. (2023). Formatted string literals. https://docs.python.org/3/tutorial/inputoutput.html
for index, row in regional_variability.iterrows():
    print(f"Region: {row['Region']}")
    print(f"  - Range:              {row['Range']:,.2f} kt")
    print(f"  - IQR:                {row['IQR']:,.2f} kt")
    print(f"  - Variance:           {row['Variance']:,.2f} kt")
    print(f"  - Standard Deviation: {row['Std_Dev']:,.2f} kt")
    print("-" * 20)
```

#### Frequency histogram for Metric Tons per Capita

```python

# This loads the libraries needed for the histogram
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

# This is the dataset being loaded into the workspace
dataset = pd.read_csv('Carbon_(CO2)_Emissions_by_Country.csv')

# This is the website referred to for creating the histogram
#https://www.geeksforgeeks.org/data-visualization/plotting-histogram-in-python-using-matplotlib/
#https://www.tutorialspoint.com/seaborn/seaborn_facetgrid_method.htm
#https://seaborn.pydata.org/generated/seaborn.FacetGrid
# This is the code to create the histogram for the "Metric Tons Per Capita" column in the dataset
# The makes sure that the individual histograms are situated in rows of 3 and the x-axis is not shared to show the full range of values for each region
grid = sns.FacetGrid(dataset, col='Region', col_wrap = 3, height = 4, aspect = 1.2, hue = 'Region', palette = 'Set2', sharex = False, sharey = False)

# This is to show a smooth density curve over the histogram bars
# This is the reference for the kde parameter in the histplot:
# Seaborn. (2025). seaborn.histplot. Seaborn. https://seaborn.pydata.org/generated/seaborn.histplot.html
grid.map(sns.histplot, 'Metric Tons Per Capita', bins=20, kde=True)

# This is for the title and axis labels of each individual histogram
# This is the reference for the axis labels and titles for the FacetGrid:
# Seaborn. (2025). seaborn.histplot. Seaborn. https://seaborn.pydata.org/generated/seaborn.FacetGrid.html#seaborn.FacetGrid.set_titles
grid.set_titles("{col_name} Region")
grid.set_axis_labels("Metric Tons Per Capita", "Frequency (Count of Records)")

# This is for the title and to adjust the layout to prevent overlap
# This is the reference for the title and layout adjustment for the FacetGrid:
# Matplotlib. (2025). matplotlib.pyplot.subplots. Matplotlib. https://matplotlib.org/stable/api/_as_gen/matplotlib.pyplot.subplots.html
plt.subplots_adjust(top=0.9)
grid.fig.suptitle('Regional Frequency Distribution of CO2 Emissions (Metric Tons Per Capita)', fontsize=16)

# This displays the final plot//graph
plt.tight_layout()
plt.show()
```
